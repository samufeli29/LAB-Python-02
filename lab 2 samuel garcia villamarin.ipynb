# Ejercicio 1 - Lambdas

```python
a = lambda n: n % 3 == 0
b = lambda n: n ** 3
c = lambda x, y: x * y
d = lambda x, y: x if x > y else y
e = lambda palabra: palabra.lower().startswith("a")
f = lambda celsius: celsius * 9 / 5 + 32

print("a) 9 múltiplo de 3:", a(9), "| 10:", a(10))
print("b) cubo de 4:", b(4))
print("c) 6 * 7:", c(6, 7))
print("d) mayor entre 3 y 8:", d(3, 8))
print("e) 'Avión' empieza con A:", e("Avión"), "| 'casa':", e("casa"))
print("f) 100 °C en °F:", f(100))
```

`e` usa `.lower()` para que sirva con "A" y "a".

# Ejercicio 2 - Listas por comprensión

```python
numeros = [5, 10, 15, 20, 25, 12, 7]
palabras = ["python", "sol", "programar", "lista", "ana", "computador"]

# a) tuplas (n, n²)
print("a)", [(n, n ** 2) for n in numeros])

# b) cubos de los mayores a 10
print("b)", [n ** 3 for n in numeros if n > 10])

# c) solo pares
print("c)", [n for n in numeros if n % 2 == 0])

# d) "valor: x"
print("d)", [f"valor: {n}" for n in numeros])

# e) Celsius -> Fahrenheit
celsius = [0, 20, 37, 100]
print("e)", [c * 9 / 5 + 32 for c in celsius])

# f) palabras de más de 5 letras
print("f)", [p for p in palabras if len(p) > 5])

# g) iniciales
print("g)", [p[0] for p in palabras])

# h) par / impar
print("h)", ["par" if n % 2 == 0 else "impar" for n in numeros])

# i) aplanar matriz
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("i)", [x for fila in matriz for x in fila])
```

En el aplanado (i) el orden de los `for` es el mismo que usarías en ciclos anidados: primero la fila, luego el elemento.

# Ejercicio 3 - Map

```python
# a) Celsius -> Fahrenheit
temps = [0, 25, 30, 100]
print("a)", list(map(lambda c: c * 9 / 5 + 32, temps)))

# b) sumar dos listas elemento a elemento
l1 = [1, 2, 3]
l2 = [10, 20, 30]
print("b)", list(map(lambda x, y: x + y, l1, l2)))

# c) áreas de círculos
import math
radios = [1, 2, 3.5]
print("c)", list(map(lambda r: round(math.pi * r ** 2, 2), radios)))

# d) (nombre, valor + 10%)
precios = {"pan": 1000, "leche": 2500, "café": 5000}
print("d)", list(map(lambda kv: (kv[0], kv[1] * 1.10), precios.items())))

# e) matriz x 10
matriz = [[1, 2, 3], [4, 5, 6]]
print("e)", list(map(lambda fila: list(map(lambda x: x * 10, fila)), matriz)))
```

`map` devuelve un iterador, por eso lo envolvemos en `list()`. En (b), `map` con dos listas recorre ambas en paralelo. En (e) usamos un `map` dentro de otro: uno por fila y otro por elemento.

# Ejercicio 4 - Filter

```python
palabras = ["sol", "casa", "arbol", "radar", "ojo", "elefante", "anilina"]

# a) más de 4 letras
print("a)", list(filter(lambda p: len(p) > 4, palabras)))

# b) quitar None
mezcla = [1, None, "a", None, 0, 3]
print("b)", list(filter(lambda x: x is not None, mezcla)))

# c) empiezan con vocal
print("c)", list(filter(lambda p: p[0].lower() in "aeiou", palabras)))

# d) palíndromos
print("d)", list(filter(lambda p: p == p[::-1], palabras)))

# e) números que terminan en 5
numeros = [15, 20, 25, 33, 105, 50]
print("e)", list(filter(lambda n: n % 10 == 5, numeros)))

# f) productos con precio > 100
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150},
]
print("f)", list(filter(lambda p: p["precio"] > 100, productos)))
```

En (b) uso `is not None` y no `if x` porque así no se pierde el `0`, que también es "falso" pero no es nulo.

# Ejercicio 4 - Reduce

```python
from functools import reduce

# a) multiplicación de todos
numeros = [2, 3, 4, 5]
print("a)", reduce(lambda acc, x: acc * x, numeros))

# b) concatenar palabras
palabras = ["Hola", " ", "mundo", "!"]
print("b)", reduce(lambda acc, p: acc + p, palabras))

# c) el mayor, con función auxiliar
def mayor(x, y):
    return x if x > y else y

print("c)", reduce(mayor, [4, 17, 9, 2, 11]))
```

`reduce` va "acumulando": toma los dos primeros elementos, los combina, y ese resultado lo combina con el siguiente, y así hasta quedar un solo valor.

# Ejercicio 4 - Sorted

```python
# a) palabras alfabéticamente
palabras = ["pera", "manzana", "uva", "banano", "kiwi"]
print("a)", sorted(palabras))

# b) tuplas por el primer elemento
tuplas = [(3, "c"), (1, "a"), (2, "b")]
print("b)", sorted(tuplas, key=lambda t: t[0]))

# c) precio de mayor a menor, y luego por nombre
productos = [("Camisa", 30), ("Pantalón", 55), ("Medias", 10), ("Chaqueta", 80), ("Gorra", 30)]
print("c)", sorted(productos, key=lambda p: (-p[1], p[0])))

# d) por longitud
print("d)", sorted(palabras, key=len))

# e) primero pares, luego impares
numeros = [5, 2, 9, 4, 7, 6, 1]
print("e)", sorted(numeros, key=lambda n: n % 2))

# f) conteo de letras ordenado por frecuencia
palabra = "banana"
conteo = {}
for letra in palabra:
    conteo[letra] = conteo.get(letra, 0) + 1
ordenado = dict(sorted(conteo.items(), key=lambda kv: kv[1], reverse=True))
print("f)", ordenado)

# g) empleados por departamento y luego por edad
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28),
]
print("g)", sorted(empleados, key=lambda e: (e[1], e[2])))
```

- En (c) el enunciado habla de diccionarios pero los datos son tuplas, así que lo resolví con tuplas. El truco es el signo `-` en el precio para invertir el orden solo de ese criterio.
- En (e) `n % 2` da 0 para pares y 1 para impares. Como `sorted` es estable, dentro de cada grupo se mantiene el orden original.
- Para ordenar por varios criterios se devuelve una tupla en `key`.

# Ejercicio 5 - Combinando funciones

```python
# a) pares y luego al cuadrado
numeros = [1, 2, 3, 4, 5, 6]
print("a)", list(map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numeros))))

# b) cubo -> filtrar > 100 -> a cadenas
nums = [2, 4, 5, 6, 3]
cubos = map(lambda x: x ** 3, nums)
grandes = filter(lambda x: x > 100, cubos)
print("b)", list(map(str, grandes)))

# c) > 10 -> cuadrado -> ordenar de mayor a menor
nums = [5, 12, 8, 20, 11, 3]
mayores = filter(lambda x: x > 10, nums)
cuadrados = map(lambda x: x ** 2, mayores)
print("c)", sorted(cuadrados, reverse=True))
```

Mismo patrón en los tres: cada función recibe lo que sale de la anterior, como una cadena de montaje. (El enunciado dice "usando sorted", pero solo el (c) lo necesita; los demás se resuelven con `map` y `filter`.)

# Ejercicio 6 - Minimax con NIM
**Reglas:** pila de piedras, cada turno se quitan 1, 2 o 3, y pierde quien no pueda mover (le toca con 0 piedras). MAX = computadora, MIN = humano, y arranca MIN.

**Cómo valoramos los estados (utilidad):** `+1` si gana la CPU, `-1` si gana el humano. Si le toca mover a MAX con 0 piedras, MAX pierde (`-1`); si le toca a MIN, MIN pierde (`+1`).

```python
def minimax(piedras, es_max):
    # Estado terminal: a quien le toca no puede mover y pierde
    if piedras == 0:
        return -1 if es_max else 1

    if es_max:
        mejor = float("-inf")
        for quitar in range(1, min(3, piedras) + 1):
            mejor = max(mejor, minimax(piedras - quitar, False))
        return mejor
    else:
        mejor = float("inf")
        for quitar in range(1, min(3, piedras) + 1):
            mejor = min(mejor, minimax(piedras - quitar, True))
        return mejor

def mejor_jugada(piedras):
    """Jugada de la CPU (MAX): la que lleva al valor minimax más alto."""
    mejor_valor, mejor_quitar = float("-inf"), 1
    for quitar in range(1, min(3, piedras) + 1):
        valor = minimax(piedras - quitar, False)   # luego le toca a MIN
        if valor > mejor_valor:
            mejor_valor, mejor_quitar = valor, quitar
    return mejor_quitar

def jugar_nim(piedras=10):
    turno_humano = True            # el primer turno es de MIN (humano)
    print(f"¡NIM! Hay {piedras} piedras. Se quitan 1, 2 o 3. Pierde quien no pueda mover.")

    while piedras > 0:
        print(f"\nPiedras: {piedras} " + "●" * piedras)
        if turno_humano:
            while True:
                try:
                    quitar = int(input("Tu turno, ¿cuántas quitas (1-3)? "))
                except ValueError:
                    print("  Escribe un número entero.")
                    continue
                if 1 <= quitar <= min(3, piedras):
                    break
                print(f"  Solo puedes quitar entre 1 y {min(3, piedras)}.")
            print(f"Tú quitas {quitar}.")
        else:
            quitar = mejor_jugada(piedras)
            print(f"La CPU quita {quitar}.")
        piedras -= quitar
        turno_humano = not turno_humano

    # Si ya no hay piedras, pierde a quien le toca mover
    if turno_humano:
        print("\nNo te quedan jugadas. ¡Gana la computadora! 🤖")
    else:
        print("\nLa CPU no puede mover. ¡Ganaste! 🎉")

jugar_nim(10)
```

**Cómo funciona:** `minimax` explora todas las partidas posibles. En los turnos de MAX elige el valor más alto, y en los de MIN asume que el humano elige el más bajo (juega lo mejor posible contra la CPU). `mejor_jugada` prueba cada movimiento de la CPU y se queda con el de mayor valor.

**Dato curioso:** con 10 piedras y empezando el humano, el que arranca gana si juega perfecto: quita 2 y deja 8, y después, cada vez que la CPU quite k, tú quitas 4 - k para dejar siempre un múltiplo de 4. Los múltiplos de 4 son las posiciones perdedoras para quien tiene el turno. Así que la CPU solo puede ganar si te equivocas, pero cuando te equivoques, no te perdona. 😄
